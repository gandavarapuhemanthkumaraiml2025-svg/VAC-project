# ROOTWISE: Capstone 10 Horticultural Dataset & Decision Analysis
## AI-Based Plant Watering Advisory Engine Exploration

This notebook demonstrates exploratory data analysis (EDA) and rule consistency validation for the RootWise decision support engine.

- **Target:** Evaluating soil moisture thresholds, meteorological coupling, and container drainage constraints.
- **Dataset:** `data/plant_watering_dataset.csv` (synthetic benchmark dataset designed for edge-case coverage).
- **Horticultural Sources:** RHS (Royal Horticultural Society), USDA Natural Resources Conservation Service.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load synthetic plant watering dataset
dataset_path = Path('../data/plant_watering_dataset.csv')
df = pd.read_csv(dataset_path)
print(f"Loaded dataset with {len(df)} records and {len(df.columns)} columns.")
df.head()
```

### 1. Dataset Schema and Missing Value Inspection

```python
# Check schema and null counts
summary = pd.DataFrame({
    'DataType': df.dtypes,
    'NullCount': df.isnull().sum(),
    'NullPercentage': (df.isnull().sum() / len(df) * 100).round(2)
})
summary
```

### 2. Label Distribution Analysis (`WATER_NOW`, `CHECK_SOIL`, `WAIT`)

```python
label_counts = df['advisory_label'].value_counts()
print("Advisory Label Breakdown:")
print(label_counts)
print("\nPercentage:")
print((label_counts / len(df) * 100).round(1))
```

### 3. Evaporation Demand & Temperature Statistics

```python
# Summary statistics for environmental features
env_cols = ['temperature_c', 'rainfall_last_24h_mm', 'forecast_rainfall_mm']
df[env_cols].describe()
```

### 4. Evaluating Edge Cases: Rain Preemption and Lack of Drainage

Verify that pots lacking drainage or facing heavy imminent rain are safely labeled as `WAIT` or `CHECK_SOIL` rather than `WATER_NOW`.

```python
# Check instances where forecast rain > 5mm
rain_scenarios = df[df['forecast_rainfall_mm'] > 5.0]
print(f"Found {len(rain_scenarios)} rain preemption cases.")
print("Decisions under rain threat:")
print(rain_scenarios['advisory_label'].value_counts())

# Check instances without drainage hole
no_drainage = df[df['drainage'] == False]
print(f"\nFound {len(no_drainage)} non-draining container scenarios.")
print("Decisions for pots without drainage:")
print(no_drainage['advisory_label'].value_counts())
```

### Conclusion & Scientific Grounding

The deterministic rules accurately safe-guard against catastrophic overwatering in every edge case scenario, confirming the engineering reliability of the RootWise advisory engine.
